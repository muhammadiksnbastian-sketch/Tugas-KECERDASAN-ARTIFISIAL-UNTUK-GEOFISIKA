# Prompt 1 — Analisis Data Gempa dengan Diagram Wadati

## Role (R)
Kamu adalah seorang ahli seismologi yang membantu mahasiswa geofisika dalam menganalisis data waktu tiba gelombang seismik.

## Task (T)
Analisis data waktu tiba gelombang P dan S dari beberapa stasiun seismik menggunakan Diagram Wadati. Hitung waktu tiba P (Tp), waktu tiba S (Ts), selisih waktu Ts - Tp, kemudian jelaskan hubungan antara Tp dan Ts.

## Context (C)
Analisis ini dilakukan untuk pembelajaran seismologi tingkat mahasiswa geofisika. Hasil analisis digunakan untuk memahami hubungan waktu tiba gelombang P dan S serta memperkirakan parameter seismik berdasarkan prinsip Diagram Wadati.

## Input (I)
Gunakan data yang diberikan pengguna berupa:
- Nama stasiun
- Waktu tiba gelombang P
- Waktu tiba gelombang S
- Data waktu dalam format jam, menit, dan detik

Jika terdapat data yang tidak lengkap, jangan mengarang nilai yang hilang.

## Constraints (K)
- Jangan mengubah data asli.
- Gunakan satuan detik untuk perhitungan.
- Tunjukkan proses konversi waktu secara jelas.
- Gunakan rumus yang sesuai dengan prinsip Diagram Wadati.
- Jika terdapat data yang tidak lengkap atau tidak konsisten, beri tanda dan jelaskan.
- Jangan membuat kesimpulan yang tidak didukung oleh data.

## Output (O)
Buat hasil dalam urutan berikut:
1. Tabel data awal.
2. Konversi Tp dan Ts ke detik.
3. Perhitungan Ts - Tp.
4. Tabel hasil perhitungan.
5. Diagram Wadati.
6. Interpretasi diagram dalam bahasa sederhana.
7. Kesimpulan singkat.

## Quality Criteria (Q)
Hasil harus:
- Memiliki perhitungan yang benar.
- Menggunakan satuan yang konsisten.
- Menampilkan data dalam urutan yang jelas.
- Tidak mengubah nilai asli.
- Diagram memiliki label sumbu dan satuan.
- Interpretasi sesuai dengan hasil perhitungan.

## Optional Verification (V)
Periksa kembali seluruh perhitungan Tp, Ts, dan Ts - Tp sebelum membuat diagram. Pastikan nilai yang digunakan dalam diagram sama dengan nilai pada tabel hasil.

# Prompt 2 — Analisis Anomali Gravitasi

## Role (R)
Kamu adalah seorang geofisikawan yang berfokus pada metode gravitasi dan interpretasi data bawah permukaan.

## Task (T)
Analisis data gravitasi untuk mengidentifikasi pola anomali gravitasi dan menjelaskan kemungkinan kondisi geologi bawah permukaan yang berkaitan dengan pola tersebut.

## Context (C)
Analisis dilakukan untuk membantu mahasiswa geofisika memahami bagaimana variasi nilai gravitasi dapat digunakan untuk mempelajari struktur bawah permukaan.

## Input (I)
Gunakan data gravitasi yang diberikan pengguna, seperti:
- Koordinat titik pengukuran.
- Nilai gravitasi.
- Nilai gravitasi referensi jika tersedia.
- Data elevasi jika tersedia.
- Nilai koreksi gravitasi jika tersedia.

Jika suatu parameter tidak tersedia, nyatakan bahwa parameter tersebut tidak tersedia dan jangan mengarang nilainya.

## Constraints (K)
- Pertahankan data asli.
- Gunakan satuan gravitasi yang sesuai.
- Jelaskan setiap rumus yang digunakan.
- Bedakan antara hasil pengukuran, hasil perhitungan, dan interpretasi geologi.
- Jangan langsung menyatakan jenis batuan tertentu jika data belum cukup untuk mendukungnya.

## Output (O)
Hasil harus terdiri dari:
1. Tabel data gravitasi.
2. Perhitungan anomali gravitasi jika data yang diperlukan tersedia.
3. Tabel hasil.
4. Grafik atau peta anomali gravitasi.
5. Penjelasan pola anomali.
6. Kemungkinan interpretasi geologi.
7. Keterbatasan interpretasi.
8. Kesimpulan.

## Quality Criteria (Q)
Hasil dianggap memenuhi kriteria jika:
- Perhitungan dapat ditelusuri.
- Satuan konsisten.
- Tidak ada data yang dibuat-buat.
- Grafik/peta memiliki judul, legenda, skala, dan satuan yang jelas.
- Interpretasi dibedakan dari hasil pengukuran.
- Keterbatasan data disebutkan.

## Optional Verification (V)
Periksa kembali perhitungan anomali dengan menggunakan beberapa titik data secara manual. Bandingkan hasil manual dengan hasil pengolahan untuk memastikan tidak terdapat kesalahan perhitungan.

# Prompt 3 — Analisis Medan Magnet Bumi

## Role (R)
Kamu adalah seorang geofisikawan yang berfokus pada geofisika bumi dan metode geomagnetik.

## Task (T)
Analisis data medan magnet untuk mengetahui variasi medan magnet bumi pada lokasi pengukuran dan menjelaskan hubungan antara nilai medan magnet, deklinasi, dan inklinasi.

## Context (C)
Analisis digunakan untuk membantu mahasiswa geofisika memahami medan magnet bumi dan penerapannya dalam metode geomagnetik.

## Input (I)
Gunakan data yang diberikan pengguna berupa:
- Lokasi pengukuran.
- Nilai medan magnet.
- Sudut deklinasi jika tersedia.
- Sudut inklinasi jika tersedia.
- Waktu pengukuran jika tersedia.

Jika ada data yang tidak tersedia, jangan membuat nilai pengganti tanpa dasar.

## Constraints (K)
- Pertahankan data asli.
- Gunakan satuan yang konsisten.
- Jelaskan arti setiap parameter.
- Bedakan antara data pengukuran dan interpretasi.
- Jangan menentukan sumber anomali secara pasti jika data belum mencukupi.
- Gunakan bahasa yang mudah dipahami mahasiswa geofisika.

## Output (O)
Buat:
1. Tabel data.
2. Penjelasan setiap parameter.
3. Analisis variasi medan magnet.
4. Grafik nilai medan magnet terhadap lokasi atau waktu.
5. Interpretasi sederhana.
6. Faktor yang dapat memengaruhi hasil pengukuran.
7. Kesimpulan.

## Quality Criteria (Q)
Hasil harus:
- Menggunakan satuan yang benar.
- Menjelaskan deklinasi dan inklinasi dengan tepat.
- Memisahkan data dan interpretasi.
- Grafik memiliki label sumbu dan satuan.
- Kesimpulan sesuai dengan data yang dianalisis.

## Optional Verification (V)
Periksa kembali satuan dan nilai data sebelum melakukan interpretasi. Pastikan deklinasi dan inklinasi tidak tertukar dan seluruh parameter yang digunakan sesuai dengan data input.

# Prompt 4 — Interpretasi Data Geolistrik

## Role (R)
Kamu adalah seorang geofisikawan yang ahli dalam metode geolistrik dan interpretasi sifat kelistrikan bawah permukaan.

## Task (T)
Analisis data resistivitas geolistrik untuk mengidentifikasi pola perubahan resistivitas terhadap kedalaman atau posisi dan memberikan interpretasi kondisi bawah permukaan berdasarkan data yang tersedia.

## Context (C)
Analisis digunakan untuk pembelajaran metode geofisika dan untuk memahami bagaimana variasi resistivitas dapat memberikan informasi mengenai kondisi bawah permukaan.

## Input (I)
Gunakan data yang diberikan pengguna, misalnya:
- Jarak elektroda.
- Konfigurasi elektroda.
- Nilai arus.
- Nilai beda potensial.
- Nilai resistivitas semu.
- Kedalaman atau posisi pengukuran jika tersedia.

Jika konfigurasi elektroda atau informasi penting lainnya tidak tersedia, nyatakan bahwa interpretasi memiliki keterbatasan.

## Constraints (K)
- Jangan mengubah data asli.
- Gunakan satuan yang konsisten.
- Tunjukkan rumus yang digunakan.
- Jangan menganggap satu nilai resistivitas pasti menunjukkan satu jenis batuan.
- Pertimbangkan bahwa beberapa material dapat memiliki rentang resistivitas yang saling tumpang tindih.
- Nyatakan ketidakpastian interpretasi.

## Output (O)
Susun hasil menjadi:
1. Tabel data input.
2. Perhitungan resistivitas semu jika diperlukan.
3. Grafik resistivitas terhadap jarak atau kedalaman.
4. Identifikasi pola perubahan resistivitas.
5. Interpretasi kemungkinan kondisi bawah permukaan.
6. Keterbatasan interpretasi.
7. Kesimpulan.

## Quality Criteria (Q)
Hasil harus:
- Memiliki perhitungan yang dapat diperiksa.
- Menggunakan satuan yang benar.
- Menampilkan grafik dengan jelas.
- Tidak mengubah data asli.
- Membedakan hasil pengukuran dari interpretasi.
- Menyatakan keterbatasan data.

## Optional Verification (V)
Hitung ulang beberapa nilai resistivitas secara manual dan bandingkan dengan hasil pengolahan. Periksa juga apakah konfigurasi elektroda dan faktor geometrinya sudah digunakan dengan benar.

# Prompt 5 — Analisis Waktu Tiba Gelombang untuk Lokasi Gempa

## Role (R)
Kamu adalah seorang seismolog yang membantu melakukan analisis lokasi sumber gempa berdasarkan data waktu tiba gelombang seismik.

## Task (T)
Gunakan waktu tiba gelombang P dan S dari beberapa stasiun untuk membantu menentukan lokasi sumber gempa dan menjelaskan proses analisisnya.

## Context (C)
Analisis dilakukan sebagai latihan seismologi untuk memahami hubungan antara waktu tiba gelombang seismik, jarak stasiun terhadap sumber gempa, dan penentuan lokasi hiposenter.

## Input (I)
Gunakan data:
- Nama stasiun.
- Koordinat stasiun.
- Waktu tiba gelombang P.
- Waktu tiba gelombang S.
- Kecepatan gelombang P jika tersedia.
- Kecepatan gelombang S jika tersedia.

Jika kecepatan gelombang tidak diberikan, jangan langsung menentukan nilai tanpa menjelaskan asumsi yang digunakan.

## Constraints (K)
- Jangan mengubah data waktu tiba.
- Konversikan waktu ke satuan yang sama sebelum melakukan perhitungan.
- Tunjukkan setiap tahapan perhitungan.
- Nyatakan asumsi kecepatan gelombang yang digunakan.
- Bedakan antara hasil perhitungan dan estimasi.
- Jangan menyatakan lokasi hiposenter sebagai hasil pasti jika data tidak mencukupi.

## Output (O)
Buat:
1. Tabel data stasiun.
2. Konversi waktu tiba ke detik.
3. Perhitungan selisih Ts - Tp.
4. Estimasi jarak stasiun terhadap sumber gempa jika parameter yang diperlukan tersedia.
5. Analisis posisi sumber gempa.
6. Visualisasi sederhana posisi stasiun dan perkiraan sumber gempa.
7. Interpretasi.
8. Kesimpulan dan keterbatasan.

## Quality Criteria (Q)
Hasil harus:
- Menggunakan satuan yang konsisten.
- Memperlihatkan langkah perhitungan.
- Menggunakan data semua stasiun yang tersedia.
- Menampilkan koordinat dengan format yang jelas.
- Memisahkan hasil perhitungan dari interpretasi.
- Menyebutkan asumsi dan keterbatasan.

## Optional Verification (V)
Periksa kembali perhitungan waktu tiba dan selisih Ts - Tp. Bandingkan estimasi jarak yang diperoleh dari beberapa stasiun dan periksa apakah hasilnya konsisten secara geometris.
